Dataframe Basic Stats

```python
def get_basic_stats(df: pd.DataFrame):
    """
    This function returns a dictionary of basic statistics for a given DataFrame.
    """
    print(f"shape: {df.shape}")
    non_id_cols = [col for col in df.columns if col not in id_cols]
    #id columns are treated as numerical values
    print(f"dataframe basic continuous statistics \n{df[non_id_cols].describe()}")
    #displays statistics like frequency
    #TODO: move this print function call elsewhere, only necessary for training data
    print(f"dataframe cat statistics \n {df[id_cols].map(lambda x: str(x)).describe()}")
```

Looking for Data Validity Issues

```python
def check_data_validity(df: pd.DataFrame):
    """
    This function checks the validity of the data in a given DataFrame by checking for missing values and unique values.
    for missing values, duplications and outliers.
    """
    print(f"missing values: \n {df.isnull().sum()}")
    print(f"unique values: \n {df.nunique()}")
    numeric_cols = []
    non_numeric_cols = []
    for col in df.columns:
        # checks if the cloumn contains numeric values and is not an id column
        if (df[col].dtype in ['int64', 'float64']) and (col not in id_cols):
            numeric_cols.append(col)
        else:
            non_numeric_cols.append(col)
    # numerical values in this case are expected to be positive
    invalid_numeric_values = df[numeric_cols].map(lambda x: 1 if x<0 else 0).sum()
    # categorical values in this case are expected to be non-empty strings or numbers
    invalid_cat_values = df[non_numeric_cols].map(lambda x: 1 if ((pd.isna(x)) or (str(x)=='')) else 0).sum()
    print(f"invalid numeric values: \n {invalid_numeric_values}")
    print(f"invalid categorical values: \n {invalid_cat_values}")

```

```python
check_data_validity(training_data)
```
