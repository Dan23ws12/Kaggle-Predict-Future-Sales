DEFINING MODELS

```python
rf0_sqrt = RandomForestRegressor(
    n_estimators=100, max_features="sqrt", n_jobs=-2, 
    min_samples_leaf=20, max_samples=0.75,
    random_state=RAND_STATE
)
```

```python
rf0_max = RandomForestRegressor(
    n_estimators=100, max_features=1, n_jobs=-2, 
    min_samples_leaf=20, max_samples=0.75,
    random_state=RAND_STATE
)
```

```python
lgb0 = lgb.LGBMRegressor(
    objective="regression",          # L2 / MSE
    metric="mse",
    n_estimators=4000,
    learning_rate=0.03,
    num_leaves=63,
    max_depth=-1,                    
    min_child_samples=100,           
    min_split_gain=0.0,
    subsample=0.8,                   
    subsample_freq=1,                
    colsample_bytree=0.9,            
    reg_lambda=3.0,                  
    reg_alpha=0.1,                   
    max_bin=250,                     
    # categorical (shop_id, item_id)
    min_data_per_group=50,
    cat_smooth=20,
    cat_l2=10,
    max_cat_to_onehot=4,
    # runtime
    n_jobs=-1,
    verbose=-1,
    random_state=RAND_STATE,
)
```

```python
xgb0 = xgb.XGBRegressor(
    objective="reg:squarederror",   # MSE
    eval_metric="rmse",
    n_estimators=4000,
    learning_rate=0.03,
    max_depth=8,
    min_child_weight=20,            # higher = less overfit on rare item_ids
    gamma=0.0,                      # raise to 0.1–1 if valid RMSE diverges
    subsample=0.8,
    colsample_bytree=0.9,           # only ~11 columns; do not drop this much
    colsample_bylevel=1.0,
    reg_lambda=3.0,
    reg_alpha=0.1,
    early_stopping_rounds=600,
    max_bin=256,
    tree_method="hist",             # accurate enough at this scale; much faster than exact
    enable_categorical=True,
    n_jobs=-1,
    random_state=RAND_STATE,
    verbosity=0,
)
```

TRAINING THE MODELS

```python
x_fit, x_eval, y_fit, y_eval = train_test_split(
    x_train0, y_train0, test_size=0.1, random_state=RAND_STATE
)
```

```python
rf0_sqrt.fit(x_train0, y_train0)
```

```python
rf0_max.fit(x_train0, y_train0)
```

```python
lgb0.fit(
    x_fit, y_fit,
    eval_set=[(x_eval, y_eval)]
)
```

```python
xgb0.fit(
    x_fit, y_fit,
    eval_set=[(x_eval, y_eval)],
    verbose=False
)
```

```python
model_performance["rf0_max"] = mean_squared_error(y_test0, rf0_max.predict(x_test0))
```

```python
model_performance["rf0_sqrt"] = mean_squared_error(y_test0, rf0_sqrt.predict(x_test0))
```

```python
model_performance["lgb0"] = mean_squared_error(y_test0, lgb0.predict(x_test0))
```

```python
model_performance["xgb0"] = mean_squared_error(y_test0, xgb0.predict(x_test0))
```
